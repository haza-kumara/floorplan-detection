Sesuai **Bagian 4** deskripsi projek, dataset mentah berada di `dataset/raw/` (berisi 121 gambar denah di `X/` dan 121 anotasi di `Y/`), sementara sampel denah terpilih untuk eksperimen disimpan di `dataset/train/`.

```python
def inspect_dataset_folder(raw_dir=DATASET_RAW_DIR, train_dir=DATASET_TRAIN_DIR):
    x_dir = os.path.join(raw_dir, "X")
    y_dir = os.path.join(raw_dir, "Y")
    x_files = sorted([f for f in os.listdir(x_dir) if f.lower().endswith(".png")]) if os.path.exists(x_dir) else []
    y_files = sorted([f for f in os.listdir(y_dir) if f.lower().endswith(".png")]) if os.path.exists(y_dir) else []
    train_files = sorted(
        [f for f in os.listdir(train_dir) if f.lower().endswith((".png", ".jpg", ".jpeg"))],
        key=lambda x: int("".join(filter(str.isdigit, x)) or 9999),
    ) if os.path.exists(train_dir) else []

    print(f"Dataset Raw (Kaggle) -> Denah (X): {len(x_files)} file PNG | Anotasi (Y): {len(y_files)} file PNG")
    print(f"Dataset Train/Uji    -> {len(train_files)} file: {train_files}")

    rows = []
    for fname in train_files:
        fpath = os.path.join(train_dir, fname)
        img = cv2.imread(fpath)
        if img is not None:
            h, w, c = img.shape
            preset = DATASET_SCALE_PRESETS.get(fname, {})
            rows.append({
                "File Uji": fname,
                "Resolusi (Px)": f"{w} x {h}",
                "Channel": c,
                "File Anotasi (Y)": preset.get("gt", "-"),
                "Keterangan": preset.get("desc", "-"),
            })
    return pd.DataFrame(rows)

dataset_summary_df = inspect_dataset_folder()
display(dataset_summary_df)

# Tampilkan 4 contoh gambar dengan karakteristik berbeda dari dataset/train
sample_names = ["denah1.png", "denah3.png", "denah7.png", "denah8.png"]
available_samples = [s for s in sample_names if os.path.exists(os.path.join(DATASET_TRAIN_DIR, s))]
if available_samples:
    fig, axes = plt.subplots(1, len(available_samples), figsize=(16, 4))
    if len(available_samples) == 1:
        axes = [axes]
    for ax, sname in zip(axes, available_samples):
        img_rgb = cv2.cvtColor(cv2.imread(os.path.join(DATASET_TRAIN_DIR, sname)), cv2.COLOR_BGR2RGB)
        ax.imshow(img_rgb)
        ax.set_title(sname, fontweight="bold")
        ax.axis("off")
    plt.suptitle("Sampel Variasi Karakteristik Gambar Denah pada Dataset", fontweight="bold")
    plt.tight_layout()
    plt.show()
```

## 4. Tahapan Pengolahan Gambar (*Image Processing* - Tahap 6)

Pada tahap ini OpenCV digunakan untuk menyiapkan citra denah agar siap dideteksi:
1. Membaca gambar dengan `cv2.imread()`
2. Konversi ke *Grayscale* dengan `cv2.cvtColor()`
3. Mengurangi *noise* dengan `cv2.GaussianBlur()`
4. Pemisahan garis dari latar belakang dengan `cv2.threshold(..., cv2.THRESH_BINARY_INV)`
5. Perbaikan struktur dinding dengan *Morphological Operations* (`erode`, `dilate`, `MORPH_CLOSE`, `MORPH_OPEN`) untuk membuang teks/angka dimensi, menyatukan arsiran dinding ganda, serta menutup celah pintu/jendela.

```python
def preprocess_floorplan(
    image_path,
    blur_ksize=3,
    thresh_val=200,
    close_ksize=9,
    open_ksize=7,
    door_gap_px=115,
    ext_gap_px=230,
):
    original = cv2.imread(image_path)
    if original is None:
        raise FileNotFoundError(f"Gambar gagal dibaca: {image_path}")

    h_img, w_img = original.shape[:2]
    scale = max(h_img, w_img) / 1950.0

    # 1. Konversi ke Grayscale
    gray = cv2.cvtColor(original, cv2.COLOR_BGR2GRAY)

    # 2. Gaussian Blur
    bk = max(3, int(blur_ksize) | 1)
    blur = cv2.GaussianBlur(gray, (bk, bk), 0)

    # 3. Thresholding Binary Inverse
    _, threshold = cv2.threshold(blur, int(thresh_val), 255, cv2.THRESH_BINARY_INV)

    # 4. Morphological Operations
    # 4a. Light opening untuk memutus garis dimensi 1px yang menempel pada huruf/angka
    k_light = max(1, min(2, int(round(2 * scale))))
    kernel_light = cv2.getStructuringElement(cv2.MORPH_RECT, (k_light, k_light))
    eroded_light = cv2.erode(threshold, kernel_light, iterations=1)
    opened_light = cv2.dilate(eroded_light, kernel_light, iterations=1)

    # Hapus komponen kecil (teks label ruangan, angka ukuran, simbol kecil)
    num_l, labels, stats, _ = cv2.connectedComponentsWithStats(opened_light, connectivity=8)
    no_text = np.zeros_like(threshold)
    min_span = max(15, int(round(45 * scale)))
    min_area = max(80, int(round(600 * scale * scale)))
    for i in range(1, num_l):
        if max(stats[i, cv2.CC_STAT_WIDTH], stats[i, cv2.CC_STAT_HEIGHT]) > min_span or stats[i, cv2.CC_STAT_AREA] > min_area:
            no_text[labels == i] = 255
    no_text = cv2.dilate(no_text, kernel_light, iterations=1)

    # 4b. Closing untuk menyatukan arsiran dinding ganda
    kc = max(3, min(11, int(round(close_ksize * scale))))
    morph_closed = cv2.morphologyEx(no_text, cv2.MORPH_CLOSE, cv2.getStructuringElement(cv2.MORPH_RECT, (kc, kc)))

    # 4c. Opening (dibatasi maks 7px agar dinding solid beresolusi tinggi tetap terjaga)
    ko = max(3, min(7, int(round(open_ksize * scale))))
    morph_opened = cv2.morphologyEx(morph_closed, cv2.MORPH_OPEN, cv2.getStructuringElement(cv2.MORPH_RECT, (ko, ko)))
    if np.count_nonzero(morph_opened) < 0.002 * (h_img * w_img):
        morph_opened = cv2.morphologyEx(morph_closed, cv2.MORPH_OPEN, np.ones((2, 2), np.uint8))

    # 4d. Ekstraksi dinding arah Horizontal & Vertikal
    kh_len = max(18, int(round(45 * scale)))
    kv_len = max(20, int(round(50 * scale)))
    kh_thk = max(2, min(6, int(round(4 * scale))))
    walls_h = cv2.morphologyEx(morph_opened, cv2.MORPH_OPEN, cv2.getStructuringElement(cv2.MORPH_RECT, (kh_len, kh_thk)))
    walls_v = cv2.morphologyEx(morph_opened, cv2.MORPH_OPEN, cv2.getStructuringElement(cv2.MORPH_RECT, (kh_thk, kv_len)))
    walls_unbridged = cv2.bitwise_or(walls_h, walls_v)

    # 4e. Penutupan celah pintu & jendela (Two-Tier Directional Closing)
    k_ext = max(6, int(round(16 * scale)))
    ext_thick = cv2.morphologyEx(morph_opened, cv2.MORPH_OPEN, cv2.getStructuringElement(cv2.MORPH_RECT, (k_ext, k_ext)))
    num_e, l_e, s_e, _ = cv2.connectedComponentsWithStats(ext_thick)
    ext_clean = np.zeros_like(ext_thick)
    for i in range(1, num_e):
        if s_e[i, cv2.CC_STAT_AREA] > int(800 * scale * scale) and max(s_e[i, cv2.CC_STAT_WIDTH], s_e[i, cv2.CC_STAT_HEIGHT]) > int(40 * scale):
            ext_clean[l_e == i] = 255

    ext_thk_bar = max(5, int(round(15 * scale)))
    ext_h = cv2.morphologyEx(ext_clean, cv2.MORPH_OPEN, cv2.getStructuringElement(cv2.MORPH_RECT, (max(15, int(round(42 * scale))), ext_thk_bar)))
    ext_v = cv2.morphologyEx(ext_clean, cv2.MORPH_OPEN, cv2.getStructuringElement(cv2.MORPH_RECT, (ext_thk_bar, max(25, int(round(75 * scale))))))
    eg = max(40, int(round(ext_gap_px * scale)))
    ext_h_b = cv2.morphologyEx(ext_h, cv2.MORPH_CLOSE, cv2.getStructuringElement(cv2.MORPH_RECT, (eg, ext_thk_bar)))
    ext_v_b = cv2.morphologyEx(ext_v, cv2.MORPH_CLOSE, cv2.getStructuringElement(cv2.MORPH_RECT, (ext_thk_bar, eg)))
    ext_all = cv2.bitwise_or(ext_clean, cv2.bitwise_or(ext_h_b, ext_v_b))

    ig = max(25, int(round(door_gap_px * scale)))
    walls_h_b = cv2.morphologyEx(walls_h, cv2.MORPH_CLOSE, cv2.getStructuringElement(cv2.MORPH_RECT, (ig, kh_thk)))
    walls_v_b = cv2.morphologyEx(walls_v, cv2.MORPH_CLOSE, cv2.getStructuringElement(cv2.MORPH_RECT, (kh_thk, ig)))
    walls_bridged = cv2.bitwise_or(cv2.bitwise_or(walls_h_b, walls_v_b), ext_all)

    # Pertahankan jaringan dinding bangunan utama (komponen terbesar)
    num_l2, labels2, stats2, _ = cv2.connectedComponentsWithStats(cv2.dilate(walls_bridged, np.ones((7, 7), np.uint8)))
    if num_l2 > 1:
        largest_label = 1 + int(np.argmax(stats2[1:, cv2.CC_STAT_AREA]))
        main_network = (labels2 == largest_label).astype(np.uint8) * 255
        walls_clean = cv2.bitwise_and(walls_bridged, main_network)
        walls_unbridged = cv2.bitwise_and(walls_unbridged, main_network)
    else:
        walls_clean = walls_bridged.copy()

    return {
        "original": original,
        "gray": gray,
        "blur": blur,
        "threshold": threshold,
        "morph_closed": morph_closed,
        "morph_opened": morph_opened,
        "walls_unbridged": walls_unbridged,
        "walls_clean": walls_clean,
        "scale_factor": scale,
    }

prep_res = preprocess_floorplan(IMAGE_PATH)
h_img, w_img, ch_img = prep_res["original"].shape
print(f"Dimensi gambar '{SELECTED_IMAGE}': {w_img} x {h_img} piksel ({ch_img} channel)")

# Visualisasi Tahapan Pengolahan Gambar menggunakan Matplotlib
fig, axes = plt.subplots(2, 3, figsize=(15, 9))
stages = [
    ("1. Gambar Asli (BGR -> RGB)", cv2.cvtColor(prep_res["original"], cv2.COLOR_BGR2RGB), None),
    ("2. Grayscale (cv2.cvtColor)", prep_res["gray"], "gray"),
    ("3. Gaussian Blur (cv2.GaussianBlur)", prep_res["blur"], "gray"),
    ("4. Thresholding (cv2.threshold)", prep_res["threshold"], "gray"),
    ("5. Morphological Closing & Opening", prep_res["morph_opened"], "gray"),
    ("6. Mask Dinding Bersih (Siap Deteksi)", prep_res["walls_clean"], "gray"),
]
for ax, (title, img_data, cmap) in zip(axes.flat, stages):
    ax.imshow(img_data, cmap=cmap)
    ax.set_title(title, fontsize=11, fontweight="bold")
    ax.axis("off")

plt.suptitle(f"SmartPlan - Tahapan Pengolahan Citra ({SELECTED_IMAGE})", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "nb_preprocessing_stages.png"), dpi=150, bbox_inches="tight")
plt.show()
```

## 5. Deteksi Dinding dan Area Ruangan (Tahap 7)

Setelah preprocessing selesai, sistem melakukan dua proses identifikasi utama:
1. **Deteksi Garis Dinding**: Penipisan (*skeletonization*) mask dinding + **Probabilistic Hough Line Transform (`cv2.HoughLinesP`)** serta penggabungan segmen sejajar agar panjang dinding tidak terhitung ganda.
2. **Deteksi Area Ruangan**: **Flood Fill (`cv2.floodFill`)** dari sudut luar gambar untuk memisahkan area eksterior bangunan, diikuti **Contour Detection (`cv2.findContours`)** untuk mengekstrak setiap area ruangan tertutup.

```python
ROOM_COLORS_BGR = [
    (180, 210, 120), (140, 180, 240), (230, 180, 130), (190, 150, 230),
    (130, 220, 235), (210, 210, 130), (160, 230, 170), (150, 160, 240),
    (225, 195, 165), (175, 220, 205), (200, 170, 210), (145, 205, 245),
]

def _skeletonize(binary_img):
    skel = np.zeros(binary_img.shape, np.uint8)
    temp_img = (binary_img > 0).astype(np.uint8) * 255
    element = cv2.getStructuringElement(cv2.MORPH_CROSS, (3, 3))
    for _ in range(100):
        eroded = cv2.erode(temp_img, element)
        temp = cv2.subtract(temp_img, cv2.dilate(eroded, element))
        skel = cv2.bitwise_or(skel, temp)
        temp_img = eroded.copy()
        if cv2.countNonZero(temp_img) == 0:
            break
    return skel

def _merge_collinear_segments(lines, tol_pos=25.0, tol_gap=25.0):
    h_segs, v_segs, d_segs = [], [], []
    for x1, y1, x2, y2 in lines:
        length = float(np.hypot(x2 - x1, y2 - y1))
        if length < 5:
            continue
        angle = float(np.abs(np.degrees(np.arctan2(y2 - y1, x2 - x1))))
        if angle <= 15.0 or angle >= 165.0:
            h_segs.append(((y1 + y2) / 2.0, float(min(x1, x2)), float(max(x1, x2))))
        elif 75.0 <= angle <= 105.0:
            v_segs.append(((x1 + x2) / 2.0, float(min(y1, y2)), float(max(y1, y2))))
        else:
            d_segs.append({"x1": int(x1), "y1": int(y1), "x2": int(x2), "y2": int(y2), "orientation": "Diagonal", "length_px": length})

    def _merge_1d(segs):
        if not segs:
            return []
        segs.sort(key=lambda s: s[0])
        groups = []
        for pos, a, b in segs:
            if not groups or abs(pos - groups[-1][0]) > tol_pos:
                groups.append([pos, [(a, b)], 1])
            else:
                cnt = groups[-1][2]
                groups[-1][0] = (groups[-1][0] * cnt + pos) / (cnt + 1)
                groups[-1][1].append((a, b))
                groups[-1][2] = cnt + 1
        merged = []
        for pos, intervals, _ in groups:
            intervals.sort(key=lambda iv: iv[0])
            cur_a, cur_b = intervals[0]
            for a, b in intervals[1:]:
                if a <= cur_b + tol_gap:
                    cur_b = max(cur_b, b)
                else:
                    merged.append((int(round(pos)), int(round(cur_a)), int(round(cur_b))))
                    cur_a, cur_b = a, b
            merged.append((int(round(pos)), int(round(cur_a)), int(round(cur_b))))
        return merged

    out = []
    for y_pos, x1, x2 in _merge_1d(h_segs):
        if x2 - x1 >= 10:
            out.append({"x1": x1, "y1": y_pos, "x2": x2, "y2": y_pos, "orientation": "Horizontal", "length_px": float(x2 - x1)})
    for x_pos, y1, y2 in _merge_1d(v_segs):
        if y2 - y1 >= 10:
            out.append({"x1": x_pos, "y1": y1, "x2": x_pos, "y2": y2, "orientation": "Vertikal", "length_px": float(y2 - y1)})
    out.extend(d_segs)
    return out

def detect_walls_hough(walls_mask, scale_factor=1.0, hough_threshold=15, min_line_length=30, max_line_gap=20):
    close_k = max(3, int(round(15 * scale_factor)))
    solid_walls = cv2.morphologyEx(walls_mask, cv2.MORPH_CLOSE, cv2.getStructuringElement(cv2.MORPH_RECT, (close_k, close_k)))
    skeleton = _skeletonize(solid_walls)

    raw_lines = cv2.HoughLinesP(
        skeleton, 1, np.pi / 180.0, threshold=hough_threshold,
        minLineLength=max(10, int(round(min_line_length * scale_factor))),
        maxLineGap=max(5, int(round(max_line_gap * scale_factor))),
    )
    lines_arr = raw_lines.reshape(-1, 4) if raw_lines is not None and len(raw_lines) > 0 else np.zeros((0, 4), dtype=np.int32)
    tol = max(10.0, 26.0 * scale_factor)
    segments = _merge_collinear_segments(lines_arr, tol_pos=tol, tol_gap=tol)

    return {
        "skeleton": skeleton,
        "segments": segments,
        "num_horizontal": sum(1 for s in segments if s["orientation"] == "Horizontal"),
        "num_vertical": sum(1 for s in segments if s["orientation"] == "Vertikal"),
        "total_wall_length_px": float(sum(s["length_px"] for s in segments)),
    }

def detect_rooms_floodfill_contours(walls_clean, scale_factor=1.0, min_room_area_ratio=0.002, max_room_area_ratio=0.65):
    h_img, w_img = walls_clean.shape[:2]
    total_img_area = float(h_img * w_img)

    walls_bordered = walls_clean.copy()
    walls_bordered[0:2, :] = 0; walls_bordered[-2:, :] = 0
    walls_bordered[:, 0:2] = 0; walls_bordered[:, -2:] = 0

    flood_filled = cv2.bitwise_not(walls_bordered)
    ff_mask = np.zeros((h_img + 2, w_img + 2), dtype=np.uint8)
    for sx, sy in [(0, 0), (w_img - 1, 0), (0, h_img - 1), (w_img - 1, h_img - 1)]:
        if flood_filled[sy, sx] > 0:
            cv2.floodFill(flood_filled, ff_mask, (sx, sy), 0)

    rk = max(5, int(round(18 * scale_factor)))
    rooms_mask = cv2.morphologyEx(flood_filled, cv2.MORPH_OPEN, cv2.getStructuringElement(cv2.MORPH_RECT, (rk, rk)))
    contours, _ = cv2.findContours(rooms_mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    raw_rooms = []
    for cnt in contours:
        area_px = float(cv2.contourArea(cnt))
        if (total_img_area * min_room_area_ratio) <= area_px <= (total_img_area * max_room_area_ratio):
            x, y, w, h = cv2.boundingRect(cnt)
            m = cv2.moments(cnt)
            cx = int(m["m10"] / m["m00"]) if m["m00"] != 0 else x + w // 2
            cy = int(m["m01"] / m["m00"]) if m["m00"] != 0 else y + h // 2
            raw_rooms.append({
                "contour": cnt, "area_px": area_px, "perimeter_px": float(cv2.arcLength(cnt, True)),
                "bbox_px": (x, y, w, h), "width_px": w, "height_px": h, "centroid_px": (cx, cy),
            })

    row_band = max(50, int(h_img * 0.12))
    raw_rooms.sort(key=lambda r: (r["centroid_px"][1] // row_band, r["centroid_px"][0]))
    rooms = []
    for idx, r in enumerate(raw_rooms, start=1):
        r["room_id"] = idx
        r["name"] = f"Ruangan {idx}"
        r["color_bgr"] = ROOM_COLORS_BGR[(idx - 1) % len(ROOM_COLORS_BGR)]
        rooms.append(r)

    return {"rooms_mask": rooms_mask, "rooms": rooms, "num_rooms": len(rooms)}

def visualize_detections(original_img, wall_res, room_res, pixels_per_meter):
    h_img, w_img = original_img.shape[:2]
    scale = max(0.5, max(h_img, w_img) / 1950.0)

    walls_vis = original_img.copy()
    for seg in wall_res["segments"]:
        color = (220, 90, 20) if seg["orientation"] == "Horizontal" else ((30, 50, 220) if seg["orientation"] == "Vertikal" else (20, 160, 20))
        cv2.line(walls_vis, (seg["x1"], seg["y1"]), (seg["x2"], seg["y2"]), color, max(2, int(round(4 * scale))))

    overlay = original_img.copy()
    for room in room_res["rooms"]:
        cv2.drawContours(overlay, [room["contour"]], -1, room["color_bgr"], -1)
    rooms_vis = cv2.addWeighted(overlay, 0.45, original_img.copy(), 0.55, 0)

    font_scale = max(0.45, 0.68 * scale)
    font_thk = max(1, int(round(2 * scale)))
    for room in room_res["rooms"]:
        cv2.drawContours(rooms_vis, [room["contour"]], -1, (20, 20, 180), max(2, int(round(3 * scale))))
        cx, cy = room["centroid_px"]
        area_m2 = room["area_px"] / (pixels_per_meter ** 2)
        l1, l2 = f"R{room['room_id']}", f"{area_m2:.2f} m2"
        (tw1, th1), _ = cv2.getTextSize(l1, cv2.FONT_HERSHEY_SIMPLEX, font_scale, font_thk)
        (tw2, th2), _ = cv2.getTextSize(l2, cv2.FONT_HERSHEY_SIMPLEX, font_scale * 0.85, font_thk)
        bw, bh = max(tw1, tw2) + int(14 * scale), th1 + th2 + int(18 * scale)
        bx1, by1 = max(5, cx - bw // 2), max(5, cy - bh // 2)
        cv2.rectangle(rooms_vis, (bx1, by1), (min(w_img - 5, bx1 + bw), min(h_img - 5, by1 + bh)), (255, 255, 255), -1)
        cv2.rectangle(rooms_vis, (bx1, by1), (min(w_img - 5, bx1 + bw), min(h_img - 5, by1 + bh)), (40, 40, 40), 1)
        cv2.putText(rooms_vis, l1, (bx1 + (bw - tw1) // 2, by1 + th1 + int(5 * scale)), cv2.FONT_HERSHEY_SIMPLEX, font_scale, (10, 10, 140), font_thk, cv2.LINE_AA)
        cv2.putText(rooms_vis, l2, (bx1 + (bw - tw2) // 2, by1 + th1 + th2 + int(12 * scale)), cv2.FONT_HERSHEY_SIMPLEX, font_scale * 0.85, (20, 20, 20), font_thk, cv2.LINE_AA)

    return walls_vis, rooms_vis

pixels_per_meter = REF_PIXELS / REF_METERS
wall_res = detect_walls_hough(prep_res["walls_unbridged"], scale_factor=prep_res["scale_factor"])
room_res = detect_rooms_floodfill_contours(prep_res["walls_clean"], scale_factor=prep_res["scale_factor"])
walls_vis, rooms_vis = visualize_detections(prep_res["original"], wall_res, room_res, pixels_per_meter)

print(f"Segmen Dinding Hough : {len(wall_res['segments'])} segmen (H: {wall_res['num_horizontal']}, V: {wall_res['num_vertical']})")
print(f"Area Ruangan         : {room_res['num_rooms']} ruangan tertutup terdeteksi")

fig, axes = plt.subplots(1, 3, figsize=(17, 6))
axes[0].imshow(cv2.cvtColor(prep_res["original"], cv2.COLOR_BGR2RGB))
axes[0].set_title("1. Denah Asli", fontweight="bold"); axes[0].axis("off")
axes[1].imshow(cv2.cvtColor(walls_vis, cv2.COLOR_BGR2RGB))
axes[1].set_title(f"2. Deteksi Dinding Hough ({wall_res['total_wall_length_px']/pixels_per_meter:.1f} m)", fontweight="bold"); axes[1].axis("off")
axes[2].imshow(cv2.cvtColor(rooms_vis, cv2.COLOR_BGR2RGB))
axes[2].set_title(f"3. Deteksi {room_res['num_rooms']} Area Ruangan", fontweight="bold"); axes[2].axis("off")
plt.suptitle(f"SmartPlan - Visualisasi Hasil Deteksi ({SELECTED_IMAGE})", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "nb_detection_comparison.png"), dpi=150, bbox_inches="tight")
plt.show()
```

## 6. Perhitungan Dimensi dan Luas Bangunan (Tahap 8)

Koordinat piksel setiap kontur ruangan dan segmen garis dinding dikonversi menjadi satuan **meter ($	ext{m}$)** dan **meter persegi ($	ext{m}^2$)** menggunakan skala `pixels_per_meter` ($	ext{piksel/meter}$).

```python
def measure_building_dimensions(room_res, wall_res, pixels_per_meter):
    px_to_m = 1.0 / float(pixels_per_meter)
    px2_to_m2 = px_to_m ** 2

    measured_rooms = []
    for r in room_res["rooms"]:
        l_px = float(max(r["width_px"], r["height_px"]))
        w_px = float(min(r["width_px"], r["height_px"]))
        area_m2 = float(r["area_px"]) * px2_to_m2
        measured_rooms.append({
            "No": r["room_id"],
            "Nama Ruangan": r["name"],
            "Panjang (px)": int(round(l_px)),
            "Lebar (px)": int(round(w_px)),
            "Luas Piksel (px2)": int(round(r["area_px"])),
            "Panjang (m)": round(l_px * px_to_m, 2),
            "Lebar (m)": round(w_px * px_to_m, 2),
            "Luas Ruangan (m2)": round(area_m2, 2),
            "centroid_px": r["centroid_px"],
        })

    rooms_df = pd.DataFrame(measured_rooms)
    total_floor_area_m2 = round(float(rooms_df["Luas Ruangan (m2)"].sum()), 2) if not rooms_df.empty else 0.0
    total_wall_length_m = round(float(wall_res["total_wall_length_px"]) * px_to_m, 2)

    return {
        "rooms_list": measured_rooms,
        "rooms_df": rooms_df.drop(columns=["centroid_px"]) if not rooms_df.empty else rooms_df,
        "total_floor_area_m2": total_floor_area_m2,
        "total_wall_length_m": total_wall_length_m,
    }

meas_res = measure_building_dimensions(room_res, wall_res, pixels_per_meter)
display(meas_res["rooms_df"])
print(f"Total Luas Lantai Bangunan : {meas_res['total_floor_area_m2']:.2f} m2")
print(f"Panjang Total Dinding      : {meas_res['total_wall_length_m']:.2f} m")
print("Catatan: Hasil perhitungan merupakan estimasi awal berbasis Computer Vision yang bergantung pada ketepatan deteksi dan skala yang diberikan.")

# Grafik Batang Luas Masing-Masing Ruangan
if not meas_res["rooms_df"].empty:
    fig, ax = plt.subplots(figsize=(10, 4))
    labels_r = [f"R{n}" for n in meas_res["rooms_df"]["No"]]
    values_r = meas_res["rooms_df"]["Luas Ruangan (m2)"]
    bars = ax.bar(labels_r, values_r, color="#2b6cb0", edgecolor="#1a365d")
    ax.set_title("Distribusi Luas Ruangan Terdeteksi (m2)", fontweight="bold")
    ax.set_xlabel("ID Ruangan")
    ax.set_ylabel("Luas (m2)")
    ax.grid(axis="y", linestyle="--", alpha=0.5)
    for bar, val in zip(bars, values_r):
        ax.text(bar.get_x() + bar.get_width() / 2.0, bar.get_height() + 0.4, f"{val:.2f}", ha="center", va="bottom", fontsize=9, fontweight="bold")
    plt.tight_layout()
    plt.show()
```

## 7. Perhitungan Kebutuhan Material Bangunan & Estimasi Biaya (Tahap 9 & 10)

1. **Keramik Lantai**: $\text{Jumlah Keping} = \lceil (\frac{\text{Total Luas Lantai}}{\text{Luas 1 Keping Keramik}}) \times (1 + \text{Faktor Cadangan}) \rceil$
2. **Cat Dinding**: $\text{Kebutuhan (Liter)} = \lceil \frac{(\text{Panjang Dinding} \times \text{Tinggi Dinding} - \text{Luas Pintu/Jendela}) \times \text{Jumlah Lapisan}}{\text{Daya Sebar Cat}} \rceil$
3. **Material Konstruksi Dinding Lanjutan (Opsional)**: Estimasi kebutuhan batu bata, semen, dan pasir.
4. **Estimasi Biaya**: Mengalikan kebutuhan setiap material dengan harga satuan dari `data/material_prices.csv` menggunakan **Pandas**.

```python
# Baca tabel referensi harga material dengan Pandas
price_df = pd.read_csv(PRICE_CSV_PATH)
print("Referensi Daftar Harga Material (data/material_prices.csv):")
display(price_df)

# --- Tahap 9: Perhitungan Material ---
tile_area_m2 = (TILE_LENGTH_CM / 100.0) * (TILE_WIDTH_CM / 100.0)
raw_tiles = meas_res["total_floor_area_m2"] / tile_area_m2
total_tiles = int(math.ceil(raw_tiles * (1.0 + TILE_WASTE_PERCENT / 100.0)))
boxes_needed = int(math.ceil(total_tiles / max(1, TILES_PER_BOX)))

gross_wall_area = meas_res["total_wall_length_m"] * WALL_HEIGHT_M * WALL_SIDES
opening_deduction = (DOOR_COUNT * 1.89 + WINDOW_COUNT * 1.20) * WALL_SIDES
net_wall_area = max(0.0, gross_wall_area - opening_deduction)
paint_liters_exact = (net_wall_area * PAINT_COATS) / PAINT_COVERAGE_M2_PER_L
paint_liters = int(math.ceil(paint_liters_exact))

# --- Tahap 10: Estimasi Biaya ---
def format_rp(val):
    return f"Rp {int(round(val)):,}".replace(",", ".")

cost_items = [
    {
        "Nama Material": f"Keramik Lantai ({int(TILE_LENGTH_CM)}x{int(TILE_WIDTH_CM)} cm)",
        "Keterangan": f"Luas {meas_res['total_floor_area_m2']:.2f} m2 + cadangan {TILE_WASTE_PERCENT:.0f}% (~{boxes_needed} dus)",
        "Jumlah Kebutuhan": total_tiles,
        "Satuan": "keping",
        "Harga Satuan (Rp)": TILE_UNIT_PRICE_IDR,
        "Total Biaya (Rp)": total_tiles * TILE_UNIT_PRICE_IDR,
    },
    {
        "Nama Material": "Cat Dinding Interior",
        "Keterangan": f"Luas dinding {net_wall_area:.2f} m2, {PAINT_COATS} lapis, {PAINT_COVERAGE_M2_PER_L:.1f} m2/L",
        "Jumlah Kebutuhan": paint_liters,
        "Satuan": "liter",
        "Harga Satuan (Rp)": PAINT_UNIT_PRICE_IDR,
        "Total Biaya (Rp)": paint_liters * PAINT_UNIT_PRICE_IDR,
    },
]

if INCLUDE_STRUCTURAL:
    single_wall_m2 = max(0.0, meas_res["total_wall_length_m"] * WALL_HEIGHT_M - (DOOR_COUNT * 1.89 + WINDOW_COUNT * 1.20))
    bricks_qty = int(math.ceil(single_wall_m2 * 70.0))
    cement_qty = int(math.ceil(single_wall_m2 * 0.45))
    sand_qty = round(single_wall_m2 * 0.045, 2)
    cost_items.extend([
        {"Nama Material": "Batu Bata Merah Press", "Keterangan": f"Pasangan dinding {single_wall_m2:.2f} m2 (@70 bh/m2)", "Jumlah Kebutuhan": bricks_qty, "Satuan": "buah", "Harga Satuan (Rp)": 900.0, "Total Biaya (Rp)": bricks_qty * 900.0},
        {"Nama Material": "Semen Portland (Sak 50 kg)", "Keterangan": f"Pasangan & plesteran {single_wall_m2:.2f} m2", "Jumlah Kebutuhan": cement_qty, "Satuan": "sak", "Harga Satuan (Rp)": 68000.0, "Total Biaya (Rp)": cement_qty * 68000.0},
        {"Nama Material": "Pasir Pasang & Plester", "Keterangan": f"Pasangan & plesteran {single_wall_m2:.2f} m2", "Jumlah Kebutuhan": sand_qty, "Satuan": "m3", "Harga Satuan (Rp)": 280000.0, "Total Biaya (Rp)": round(sand_qty * 280000.0, 0)},
    ])

cost_df = pd.DataFrame(cost_items)
cost_display = cost_df.copy()
cost_display["Harga Satuan"] = cost_display["Harga Satuan (Rp)"].apply(format_rp)
cost_display["Total Biaya"] = cost_display["Total Biaya (Rp)"].apply(format_rp)

print("\nTabel Kebutuhan Material dan Estimasi Biaya:")
display(cost_display[["Nama Material", "Keterangan", "Jumlah Kebutuhan", "Satuan", "Harga Satuan", "Total Biaya"]])

total_cost_idr = cost_df["Total Biaya (Rp)"].sum()
print(f">>> TOTAL ESTIMASI BIAYA MATERIAL : {format_rp(total_cost_idr)} <<<")
print("Catatan Batasan: Estimasi belum mencakup biaya tenaga kerja, transportasi, fondasi, atap, serta instalasi listrik/air.")

# Simpan tabel CSV ke notebooks/outputs/
meas_res["rooms_df"].to_csv(os.path.join(OUTPUT_DIR, "nb_rooms_summary.csv"), index=False)
cost_df.to_csv(os.path.join(OUTPUT_DIR, "nb_material_cost_summary.csv"), index=False)

# Visualisasi Proporsi Biaya Material
fig, ax = plt.subplots(figsize=(7, 4))
ax.barh(cost_df["Nama Material"], cost_df["Total Biaya (Rp)"] / 1e6, color="#2f855a")
ax.set_xlabel("Estimasi Biaya (Juta Rp)")
ax.set_title("Proporsi Estimasi Biaya per Jenis Material", fontweight="bold")
ax.invert_yaxis()
ax.grid(axis="x", linestyle="--", alpha=0.5)
for i, v in enumerate(cost_df["Total Biaya (Rp)"]):
    ax.text(v / 1e6 + 0.2, i, format_rp(v), va="center", fontsize=9)
plt.xlim(0, (cost_df["Total Biaya (Rp)"].max() / 1e6) * 1.35)
plt.tight_layout()
plt.show()
```

## 8. Pengujian dan Evaluasi Sistem (Tahap 13)

Evaluasi dilakukan melalui 3 cara:
1. Membandingkan luas hasil deteksi OpenCV terhadap **ukuran luas tertulis pada denah asli** (untuk `denah1.png`).
2. Membandingkan mask area ruangan OpenCV terhadap **mask anotasi Ground Truth** pada `dataset/raw/Y/` (metrik IoU, Precision, Recall, dan Kemiripan Luas).
3. Menjalankan pengujian lintas seluruh **10 gambar denah** pada `dataset/train/` (`denah1.png` s/d `denah10.png`).

```python
# 1. Evaluasi terhadap ukuran referensi tertulis pada denah1.png
if SELECTED_IMAGE.lower() == "denah1.png":
    ref_rooms_d1 = [
        {"nama_acuan": "Chambre 2 (Kamar Tidur 2)", "luas_acuan_m2": 10.20, "target_centroid": (738, 330)},
        {"nama_acuan": "Chambre 3 (Kamar Tidur 3)", "luas_acuan_m2": 10.02, "target_centroid": (1030, 325)},
        {"nama_acuan": "Bains (Kamar Mandi)", "luas_acuan_m2": 6.31, "target_centroid": (740, 540)},
        {"nama_acuan": "Bureau (Ruang Kerja)", "luas_acuan_m2": 7.80, "target_centroid": (1145, 525)},
        {"nama_acuan": "Chambre 1 (Kamar Tidur 1)", "luas_acuan_m2": 10.95, "target_centroid": (697, 776)},
        {"nama_acuan": "Dressing (Ruang Ganti)", "luas_acuan_m2": 2.94, "target_centroid": (871, 869)},
        {"nama_acuan": "Salon / Sejour / Labo", "luas_acuan_m2": 44.38, "target_centroid": (1240, 700)},
        {"nama_acuan": "Garage (Garasi)", "luas_acuan_m2": 32.88, "target_centroid": (815, 1175)},
    ]
    eval_rows = []
    for ref in ref_rooms_d1:
        tx, ty = ref["target_centroid"]
        best = min(meas_res["rooms_list"], key=lambda r: np.hypot(r["centroid_px"][0] - tx, r["centroid_px"][1] - ty))
        err_pct = abs(best["Luas Ruangan (m2)"] - ref["luas_acuan_m2"]) / ref["luas_acuan_m2"] * 100.0
        eval_rows.append({
            "Ruangan Acuan": ref["nama_acuan"],
            "ID Terdeteksi": f"R{best['No']}",
            "Luas Referensi (m2)": ref["luas_acuan_m2"],
            "Luas OpenCV (m2)": best["Luas Ruangan (m2)"],
            "Error (%)": round(err_pct, 2),
            "Akurasi (%)": round(max(0.0, 100.0 - err_pct), 2),
        })
    eval_d1_df = pd.DataFrame(eval_rows)
    print("A. Evaluasi Perbandingan Luas terhadap Angka Tertulis pada denah1.png:")
    display(eval_d1_df)
    tot_ref = eval_d1_df["Luas Referensi (m2)"].sum()
    tot_det = eval_d1_df["Luas OpenCV (m2)"].sum()
    tot_acc = (1.0 - abs(tot_det - tot_ref) / tot_ref) * 100.0
    print(f"Rata-rata Akurasi Per Ruangan : {eval_d1_df['Akurasi (%)'].mean():.2f}%")
    print(f"Akurasi Total Luas Ruangan    : {tot_acc:.2f}% (OpenCV {tot_det:.2f} m2 vs Referensi {tot_ref:.2f} m2)")

# 2. Evaluasi Mask Ruangan OpenCV vs Anotasi Ground Truth (dataset/raw/Y)
def eval_ground_truth_mask(pred_mask, gt_path):
    if not gt_path or not os.path.exists(gt_path):
        return None
    gt_bgr = cv2.imread(gt_path)
    if gt_bgr.shape[:2] != pred_mask.shape[:2]:
        gt_bgr = cv2.resize(gt_bgr, (pred_mask.shape[1], pred_mask.shape[0]), interpolation=cv2.INTER_NEAREST)
    b, g, r = cv2.split(gt_bgr)
    gt_room = (b > 110) & (b < 175) & (g > 110) & (g < 175) & (r < 40)
    if gt_room.sum() < 0.02 * (gt_bgr.shape[0] * gt_bgr.shape[1]):
        gt_room = (b > 40) & (b < 100) & (g > 110) & (g < 175) & (r > 210)
    pred_room = pred_mask > 0
    inter = float(np.logical_and(gt_room, pred_room).sum())
    union = float(np.logical_or(gt_room, pred_room).sum())
    gt_px, pr_px = float(gt_room.sum()), float(pred_room.sum())
    return {
        "iou": round(inter / union * 100.0, 2) if union > 0 else 0.0,
        "precision": round(inter / pr_px * 100.0, 2) if pr_px > 0 else 0.0,
        "recall": round(inter / gt_px * 100.0, 2) if gt_px > 0 else 0.0,
        "area_sim": round(max(0.0, (1.0 - abs(pr_px - gt_px) / gt_px) * 100.0), 2) if gt_px > 0 else 0.0,
        "gt_bgr": gt_bgr,
    }

gt_eval = eval_ground_truth_mask(room_res["rooms_mask"], GT_PATH)
if gt_eval:
    print(f"\nB. Evaluasi Mask terhadap Ground Truth ({os.path.basename(GT_PATH)}):")
    print(f"   IoU = {gt_eval['iou']}% | Precision = {gt_eval['precision']}% | Recall = {gt_eval['recall']}% | Kemiripan Total Luas = {gt_eval['area_sim']}%")
    fig, axes = plt.subplots(1, 2, figsize=(12, 5))
    axes[0].imshow(cv2.cvtColor(gt_eval["gt_bgr"], cv2.COLOR_BGR2RGB))
    axes[0].set_title(f"Anotasi Ground Truth ({os.path.basename(GT_PATH)})", fontweight="bold"); axes[0].axis("off")
    axes[1].imshow(room_res["rooms_mask"], cmap="gray")
    axes[1].set_title("Mask Ruangan Hasil OpenCV", fontweight="bold"); axes[1].axis("off")
    plt.tight_layout()
    plt.show()
```

```python
# 3. Benchmark Otomatis pada Seluruh 10 Gambar di dataset/train/
bench_rows = []
for fname, info in DATASET_SCALE_PRESETS.items():
    fpath = os.path.join(DATASET_TRAIN_DIR, fname)
    if not os.path.exists(fpath):
        continue
    p_res = preprocess_floorplan(fpath)
    ppm = info["ref_pixels"] / info["ref_meters"]
    w_res = detect_walls_hough(p_res["walls_unbridged"], scale_factor=p_res["scale_factor"])
    r_res = detect_rooms_floodfill_contours(p_res["walls_clean"], scale_factor=p_res["scale_factor"])
    m_res = measure_building_dimensions(r_res, w_res, ppm)
    g_res = eval_ground_truth_mask(r_res["rooms_mask"], os.path.join(DATASET_RAW_DIR, "Y", info["gt"]))
    bench_rows.append({
        "File Denah": fname,
        "Skala (px/m)": round(ppm, 1),
        "Ruangan": r_res["num_rooms"],
        "Total Luas Lantai (m2)": m_res["total_floor_area_m2"],
        "Panjang Dinding (m)": m_res["total_wall_length_m"],
        "IoU vs GT (%)": g_res["iou"] if g_res else None,
        "Kemiripan Luas vs GT (%)": g_res["area_sim"] if g_res else None,
    })

bench_df = pd.DataFrame(bench_rows)
print("C. Pengujian Lintas 10 Denah Berbeda (dataset/train/):")
display(bench_df)
bench_df.to_csv(os.path.join(OUTPUT_DIR, "nb_benchmark_10_denah.csv"), index=False)
```
